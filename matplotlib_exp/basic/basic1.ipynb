```python
import pandas as pd

df = pd.DataFrame({
    'month': [1,2,3,4,5,6,],
    'sales': [120, 135, 128, 160, 175, 190]
})

plt.plot(df['month'], df['sales'])
plt.show()
```

```python
month = [ 1, 2, 3, 4, 5, 6 ]

product_a = [ 120, 135, 128, 160, 175, 190 ]
product_b = [ 90, 110, 125, 120, 150, 165 ]

plt.plot(month, product_a)
plt.plot(month, product_b)

plt.show()
```

```python
month = [ 1, 2, 3, 4, 5, 6 ]

product_a = [ 120, 135, 128, 160, 175, 190 ]
product_b = [ 90, 110, 125, 120, 150, 165 ]

plt.plot(month, product_a, label='Product A')
plt.plot(month, product_b, label='Product B')

plt.legend()

plt.show()
```

```python
month = [ 1, 2, 3, 4, 5, 6 ]

product_a = [ 120, 135, 128, 160, 175, 190 ]
product_b = [ 90, 110, 125, 120, 150, 165 ]
product_c = [ 70, 85, 95, 110, 125, 140 ]

plt.plot(month, product_a, label='Product A')
plt.plot(month, product_b, label='Product B')
plt.plot(month, product_c, label='Product C')

plt.legend()

plt.show()
```

```python
month = [ 1, 2, 3, 4, 5, 6 ]

product_a = [ 120, 135, 128, 160, 175, 190 ]

title = 'Monthly Sales'

plt.plot(month, product_a)
plt.title(
    title,
    fontsize=18,
    # loc='left',
)

plt.xlabel('Month')
plt.ylabel('Sales')

print(plt.gca().get_title())
# print(type(plt.gca()))

print(plt.gca().get_xlabel())
print(plt.gca().get_ylabel())

plt.show()
```

```python
month = [ 1, 2, 3, 4, 5, 6 ]

product_a = [ 120, 135, 128, 160, 175, 190 ]

plt.plot(month, product_a, marker='o')

plt.xlim(0, 12)
plt.ylim(90, 200)

plt.show()
```

```python
month = [ 1, 2, 3, 4, 5, 6 ]

product_a = [ 120, 135, 128, 160, 175, 190 ]

plt.plot(month, product_a)

plt.xticks(
    [1,2,3,4],
    ['January', 'February', 'March', 'April'],
    rotation=45,
)

plt.show()
```
