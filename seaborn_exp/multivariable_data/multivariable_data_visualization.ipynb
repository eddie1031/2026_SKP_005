```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
rng = np.random.default_rng(42)

n = 150

advertising = rng.normal(50,12,n)
visitors = 400 + 8 * advertising + rng.normal(0,80,n)

conversion_rate = rng.normal(0.12, 0.02, n)
buyers = visitors * conversion_rate + rng.normal(0,5,n)

sales = 20 + 0.7 * advertising + 0.35 * buyers + rng.normal(0,8,n)
satisfaction = 3.0 + 0.006 * sales + rng.normal(0, 0.35, n)

channel = rng.choice(
    ['Online', 'Offline'],
    size=n
)

customer_df = pd.DataFrame({
    'advertising': advertising.round(1),
    'visitors': visitors.round(0).astype(int),
    'buyers': buyers.round(0).astype(int),
    'sales': sales.round(1),
    'statisfaction': np.clip(
        satisfaction,
        1,
        5,
    ).round(2),
    'channel': channel,
})

customer_df.head()
```

```python
matrix = pd.DataFrame(
    [
        [10,20,30,40],
        [15,25,35,45],
        [20,30,40,50],
        [25,35,45,55],
    ],
    index=['A','B','C','D'],
    columns=['W','X','Y','Z']
)

matrix
```

```python
sns.heatmap(data=matrix)

plt.show()
```

```python
sns.heatmap(
    data=matrix,
    cbar=False,
)

plt.show()
```

```python
sns.heatmap(
    data=matrix,
    cbar=False,
    annot=True,
    linewidths=0.5
)

plt.show()
```
