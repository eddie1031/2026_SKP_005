```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales_df = pd.DataFrame({
    "month": [
        1, 2, 3, 4, 5, 6,
        1, 2, 3, 4, 5, 6,
        1, 2, 3, 4, 5, 6,
    ],
    "region": [
        "Seoul", "Seoul", "Seoul",
        "Seoul", "Seoul", "Seoul",
        "Busan", "Busan", "Busan",
        "Busan", "Busan", "Busan",
        "Daejeon", "Daejeon", "Daejeon",
        "Daejeon", "Daejeon", "Daejeon",
    ],
    "sales": [
        120, 138, 155, 171, 188, 205,
        105, 118, 132, 149, 163, 178,
        92, 106, 119, 133, 148, 162,
    ],
    "advertising": [
        10, 12, 15, 18, 21, 24,
        9, 11, 13, 16, 19, 22,
        7, 9, 12, 14, 17, 20,
    ],
    "visitors": [
        800, 900, 1010, 1130, 1260, 1400,
        720, 810, 900, 1010, 1120, 1250,
        650, 730, 820, 910, 1020, 1140,
    ],
})
```

```python
sales_df.head()
```

```python
sns.set_theme()
```

```python
sns.lineplot(
    data=sales_df,
    x='month',
    y='sales',
    hue='region',
    marker='o'
)
```
